```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("all_scenarios.csv")

df["scenario"] = df["scenario"].replace({
    "SSP1-2.6": "Low Emissions",
    "SSP2-4.5": "Moderate Emissions",
    "SSP5-8.5": "High Emissions"
})

df["decade"] = (df["year"] // 10) * 10

warming_rates = []

for (scenario, decade), group in df.groupby(["scenario", "decade"]):
    group = group.sort_values("year")

    if len(group) < 2:
        continue

    slope = np.polyfit(
        group["year"],
        group["global_temp_c"],
        1
    )[0]

    warming_rates.append({
        "scenario": scenario,
        "decade": str(decade) + "s",
        "warming_rate": slope * 10
    })

warming_df = pd.DataFrame(warming_rates)

warming_pivot = warming_df.pivot(
    index="decade",
    columns="scenario",
    values="warming_rate"
)

scenario_order = [
    "Low Emissions",
    "Moderate Emissions",
    "High Emissions"
]

warming_pivot = warming_pivot[scenario_order]

plt.figure(figsize=(12, 6))

x = np.arange(len(warming_pivot.index))
width = 0.25

for i, scenario in enumerate(scenario_order):
    plt.bar(
        x + (i - 1) * width,
        warming_pivot[scenario],
        width,
        label=scenario
    )

plt.xlabel("Decade")
plt.ylabel("Warming Rate (°C per decade)")
plt.title("Global Warming Rate by Decade and Climate Scenario")

plt.xticks(x, warming_pivot.index)
plt.legend(title="Climate Scenario")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
high = df[df["scenario"] == "High Emissions"]
low = df[df["scenario"] == "Low Emissions"]

comparison = pd.merge(
    high[["year", "global_temp_c"]],
    low[["year", "global_temp_c"]],
    on="year",
    suffixes=("_high", "_low")
)

comparison["temperature_difference"] = (
    comparison["global_temp_c_high"] -
    comparison["global_temp_c_low"]
)

plt.figure(figsize=(12, 6))

plt.plot(
    comparison["year"],
    comparison["temperature_difference"],
    linewidth=2
)

plt.xlabel("Year")
plt.ylabel("Temperature Difference (°C)")
plt.title("Global Temperature Difference Between High and Low Emissions")
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
df = df.sort_values(["scenario", "year"])

# Year-to-year warming rate
df["warming_rate"] = df.groupby("scenario")["global_temp_c"].diff()

# Change in warming rate
df["acceleration"] = df.groupby("scenario")["warming_rate"].diff()

df["decade"] = (df["year"] // 10) * 10

# Average acceleration for each decade
acceleration_df = (
    df.groupby(["scenario", "decade"])["acceleration"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 6))

scenarios = ["Low Emissions", "Moderate Emissions", "High Emissions"]

for scenario in scenarios:
    subset = acceleration_df[
        acceleration_df["scenario"] == scenario
    ]

    plt.plot(
        subset["decade"],
        subset["acceleration"],
        marker="o",
        linewidth=2,
        label=scenario
    )

plt.axhline(0, color="black", linestyle="--", linewidth=1)

plt.xlabel("Decade")
plt.ylabel("Average Warming Acceleration")
plt.title("Average Global Warming Acceleration by Decade")
plt.xticks(rotation=45)
plt.legend(title="Climate Scenario")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
